```python
### Acessing rows using iloc and loc
import pandas as pd
data = {'Team': ['Redbull', 'Ferrai', 'McLaren'],
        'Driver': ['Max Verstappen', 'Hamilton', 'Lando Norris'],
        'Tyre': ['hard', 'soft', 'soft'],
        'Power_Unit': ['Honda', 'Ferrari', 'Mercedes']}
df = pd.DataFrame(data)
print(df)
df.iloc[0:2] #Access row by position using iloc
```

```python
### Acessing rows using iloc and loc
import pandas as pd
data = {'Team': ['Redbull', 'Ferrai', 'McLaren'],
        'Driver': ['Max Verstappen', 'Hamilton', 'Lando Norris'],
        'Tyre': ['hard', 'soft', 'soft'],
        'Power_Unit': ['Honda', 'Ferrari', 'Mercedes']}
df = pd.DataFrame(data)
print(df)
df.loc[0:2] ##access row by label using loc
```
