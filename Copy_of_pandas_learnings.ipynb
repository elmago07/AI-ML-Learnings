```python
import pandas as pd
###creating a series from a list###
data = [1,2,3,4,5]
series = pd.Series(data)
print(series)
```

```python
####Data FRame###

data = {'Team': ['Redbull', 'Ferrai', 'McLaren'],
        'Driver': ['Max Verstappen', 'Hamilton', 'Lando Norris']}
df = pd.DataFrame(data)
print(df)
```
